## 🎯 Step 4: Evaluate Trained Model (run after training completes)


```python
# ============================================================================
# AUTOMATIC EVALUATION - Test your trained model!
# ============================================================================

print("="*80)
print("🎯 Evaluating Trained Model")
print("="*80)

import sys
import torch
import numpy as np
from pathlib import Path
import yaml
import glob

# Add src to path
sys.path.insert(0, str(Path.cwd() / 'src'))

from src.models.balatro_network import create_model
from src.environment.balatro_env import BalatroEnv

# Find latest checkpoint
print("\n📂 Looking for checkpoint...")
checkpoint_files = glob.glob('checkpoints/a100/*/final_model.pt')
if not checkpoint_files:
    checkpoint_files = glob.glob('checkpoints/a100/**/checkpoint_*.pt')

if not checkpoint_files:
    print("❌ No checkpoint found! Training may still be running.")
    print("   Checkpoints are saved to: checkpoints/a100/")
else:
    checkpoint_path = max(checkpoint_files, key=lambda x: Path(x).stat().st_mtime)
    print(f"✅ Found: {checkpoint_path}")
    
    # Load model
    print("\n🤖 Loading model...")
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    with open('configs/a100.yaml') as f:
        config = yaml.safe_load(f)
    
    model = create_model(config['model']).to(device)
    checkpoint = torch.load(checkpoint_path, map_location=device)
    model.load_state_dict(checkpoint['model_state_dict'])
    model.eval()
    print("✅ Model loaded")
    
    # Quick evaluation
    print(f"\n🎮 Running {config['evaluation']['n_eval_episodes']} test episodes...")
    print("="*80)
    
    env = BalatroEnv(
        max_steps=config['environment']['max_steps'],
        starting_deck_size=config['environment']['starting_deck_size']
    )
    
    rewards = []
    lengths = []
    
    with torch.no_grad():
        for ep in range(config['evaluation']['n_eval_episodes']):
            obs, _ = env.reset()
            done = False
            ep_reward = 0
            ep_length = 0
            
            while not done:
                obs_tensor = {k: torch.FloatTensor(v).unsqueeze(0).to(device) 
                            for k, v in obs.items()}
                action, _, _, _ = model.get_action_and_value(obs_tensor, deterministic=True)
                obs, reward, terminated, truncated, _ = env.step(
                    {k: v.cpu().numpy()[0] for k, v in action.items()}
                )
                done = terminated or truncated
                ep_reward += reward
                ep_length += 1
            
            rewards.append(ep_reward)
            lengths.append(ep_length)
            print(f"   Episode {ep+1}: Reward={ep_reward:.2f}, Length={ep_length}")
    
    print("="*80)
    print(f"\n📊 Results:")
    print(f"   Mean Reward: {np.mean(rewards):.2f} ± {np.std(rewards):.2f}")
    print(f"   Mean Length: {np.mean(lengths):.1f} ± {np.std(lengths):.1f}")
    print(f"   Best: {np.max(rewards):.2f} | Worst: {np.min(rewards):.2f}")
    print("="*80)
```

```python
# Evaluate on multiple episodes
eval_env = BalatroEnv(
    max_steps=config['environment']['max_steps'],
    starting_deck_size=config['environment']['starting_deck_size']
)

n_eval_episodes = config['evaluation']['n_eval_episodes']
eval_rewards = []
eval_lengths = []
eval_max_scores = []

model.eval()
print(f"Evaluating model over {n_eval_episodes} episodes...")

with torch.no_grad():
    for episode in range(n_eval_episodes):
        obs, info = eval_env.reset()
        done = False
        episode_reward = 0
        episode_length = 0
        max_score = 0
        
        while not done:
            # Convert observation to tensor
            obs_tensor = {k: torch.FloatTensor(v).unsqueeze(0).to(device) for k, v in obs.items()}
            
            # Get action (deterministic)
            action, _, _, _ = model.get_action_and_value(obs_tensor, deterministic=True)
            
            # Step environment
            obs, reward, terminated, truncated, info = eval_env.step(
                {k: v.cpu().numpy()[0] for k, v in action.items()}
            )
            
            done = terminated or truncated
            episode_reward += reward
            episode_length += 1
            max_score = max(max_score, info.get('score', 0))
        
        eval_rewards.append(episode_reward)
        eval_lengths.append(episode_length)
        eval_max_scores.append(max_score)
        
        print(f"  Episode {episode+1}/{n_eval_episodes}: "
              f"Reward={episode_reward:.2f}, Length={episode_length}, Max Score={max_score}")

print(f"\nEvaluation Results:")
print(f"  Mean Reward: {np.mean(eval_rewards):.2f} ± {np.std(eval_rewards):.2f}")
print(f"  Mean Length: {np.mean(eval_lengths):.1f} ± {np.std(eval_lengths):.1f}")
print(f"  Mean Max Score: {np.mean(eval_max_scores):.1f} ± {np.std(eval_max_scores):.1f}")
print(f"  Best Episode Reward: {np.max(eval_rewards):.2f}")
print(f"  Worst Episode Reward: {np.min(eval_rewards):.2f}")
```
