## 🎯 Step 2: Start Training (this is where the magic happens!)


````python
# ============================================================================
# TRAINING WITH LIVE PROGRESS - The main event!
# ============================================================================
print("\n" + "="*80)
print("⚡ Starting BalatroNN Training on A100 (V3 Rewards!)")
print("="*80)

import yaml
import subprocess
import time
import re
import json
import os
from pathlib import Path
from IPython.display import display, HTML, clear_output
import threading

# Load Google Drive paths (if available)
try:
    with open('/content/balatroNN/drive_paths.json', 'r') as f:
        drive_paths = json.load(f)
    checkpoint_dir_base = drive_paths['checkpoint_dir']
    logs_dir_base = drive_paths['logs_dir']
    print(f"💾 Using Google Drive for checkpoints: {checkpoint_dir_base}")
except (FileNotFoundError, json.JSONDecodeError):
    checkpoint_dir_base = 'checkpoints/a100_aggressive'
    logs_dir_base = 'logs/a100_aggressive'
    print(f"💾 Using local storage for checkpoints: {checkpoint_dir_base}")

config_path = 'configs/a100_aggressive.yaml'
with open(config_path) as f:
    config = yaml.safe_load(f)

# Override config paths to use Google Drive
config['checkpoints']['dir'] = checkpoint_dir_base
config['training']['log_dir'] = logs_dir_base

# Create directories if they don't exist
os.makedirs(checkpoint_dir_base, exist_ok=True)
os.makedirs(logs_dir_base, exist_ok=True)

# Save modified config temporarily
temp_config_path = '/tmp/a100_aggressive_gdrive.yaml'
with open(temp_config_path, 'w') as f:
    yaml.dump(config, f)

config_path = temp_config_path

print(f"""
📋 Full Training Configuration:
   • Model: {config['model']['d_model']}d, {config['model']['num_layers']} layers ({config['model']['num_heads']} heads)
   • Batch Size: {config['training']['batch_size']:,} (MASSIVE!)
   • Total Timesteps: {config['training']['total_timesteps']:,} (50M!)
   • Expected Duration: ~15-20 hours
   • Curriculum Learning: {'✅ ENABLED' if config.get('curriculum', {}).get('enabled', False) else '❌ Disabled'}
   
💣 V3 Exponential Skip Penalties + 🎓 Curriculum Learning!
   Skip penalties double each time + Progressive joker introduction (7 phases)
""")

# Check for resume
checkpoint_dir = Path(config['checkpoints']['dir'])
resume_path = None
if checkpoint_dir.exists():
    checkpoints = sorted(checkpoint_dir.glob('*.pt'), key=lambda x: x.stat().st_mtime)
    if checkpoints:
        resume_path = str(checkpoints[-1])
        print(f"✅ Resuming from: {resume_path}\n")
    else:
        print("📌 Starting fresh training\n")
else:
    print("📌 Starting fresh training\n")

print("="*80)
print("🚀 Training started! Live updates every 30 seconds...")
print("="*80 + "\n")

# Build command
cmd = ['python', '-u', 'train.py', '--config', config_path]
if resume_path:
    cmd.extend(['--resume', resume_path])

# Start training with live output
process = subprocess.Popen(
    cmd,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    universal_newlines=True,
    bufsize=1
)

# Track metrics
last_update = time.time()
update_interval = 30  # seconds

try:
    for line in process.stdout:
        print(line, end='')  # Print all output
        
        # Parse for interesting metrics
        if any(keyword in line for keyword in ['reward', 'loss', 'timesteps', 'fps']):
            current_time = time.time()
            if current_time - last_update >= update_interval:
                print("\n" + "─"*80)
                print(f"📊 Update at {time.strftime('%H:%M:%S')}")
                print("─"*80)
                last_update = current_time
                
    process.wait()
    
    if process.returncode == 0:
        print("\n" + "="*80)
        print("✅ Training Complete!")
        print("="*80)
    else:
        print("\n" + "="*80)
        print(f"⚠️ Training ended with code {process.returncode}")
        print("="*80)
        
except KeyboardInterrupt:
    print("\n⚠️ Training interrupted by user")
    process.terminate()
    process.wait()

````

## 🎯 Step 4: Evaluate Trained Model (run after training completes)


````python
# ============================================================================
# AUTOMATIC EVALUATION - Test your trained model!
# ============================================================================

print("="*80)
print("🎯 Evaluating Trained Model")
print("="*80)

import sys
import torch
import numpy as np
from pathlib import Path
import yaml
import glob
import json

# Add src to path
sys.path.insert(0, str(Path.cwd() / 'src'))

from src.models.balatro_network import create_model
from src.environment.balatro_env import BalatroEnv

# Load Google Drive paths (if available)
try:
    with open('/content/balatroNN/drive_paths.json', 'r') as f:
        drive_paths = json.load(f)
    checkpoint_dir_base = drive_paths['checkpoint_dir']
    print(f"💾 Looking for checkpoints in Google Drive: {checkpoint_dir_base}")
except (FileNotFoundError, json.JSONDecodeError):
    checkpoint_dir_base = 'checkpoints/a100_aggressive'
    print(f"💾 Looking for checkpoints locally: {checkpoint_dir_base}")

# Find latest checkpoint
print("\n📂 Searching for checkpoint files...")
checkpoint_files = glob.glob(f'{checkpoint_dir_base}/*.pt')
if not checkpoint_files:
    checkpoint_files = glob.glob(f'{checkpoint_dir_base}/**/checkpoint_*.pt', recursive=True)
    
# Also check local directory as fallback
if not checkpoint_files:
    print("   No checkpoints in primary location, checking local...")
    checkpoint_files = glob.glob('checkpoints/a100_aggressive/*.pt')
    if not checkpoint_files:
        checkpoint_files = glob.glob('checkpoints/a100_aggressive/**/checkpoint_*.pt', recursive=True)

if not checkpoint_files:
    print("❌ No checkpoint found! Training may still be running.")
    print("   Checkpoints are saved to: checkpoints/a100_aggressive/")
else:
    checkpoint_path = max(checkpoint_files, key=lambda x: Path(x).stat().st_mtime)
    print(f"✅ Found: {checkpoint_path}")
    
    # Load model
    print("\n🤖 Loading model...")
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    with open('configs/a100_aggressive.yaml') as f:
        config = yaml.safe_load(f)
    
    model = create_model(config['model']).to(device)
    checkpoint = torch.load(checkpoint_path, map_location=device)
    model.load_state_dict(checkpoint['model_state_dict'])
    model.eval()
    print("✅ Model loaded")
    
    # Quick evaluation
    print(f"\n🎮 Running {config['evaluation']['n_eval_episodes']} test episodes...")
    print("="*80)
    
    env = BalatroEnv(
        max_steps=config['environment']['max_steps'],
        starting_deck_size=config['environment']['starting_deck_size']
    )
    
    rewards = []
    lengths = []
    
    with torch.no_grad():
        for ep in range(config['evaluation']['n_eval_episodes']):
            obs, _ = env.reset()
            done = False
            ep_reward = 0
            ep_length = 0
            
            while not done:
                obs_tensor = {k: torch.FloatTensor(v).unsqueeze(0).to(device) 
                            for k, v in obs.items()}
                action, _, _, _ = model.get_action_and_value(obs_tensor, deterministic=True)
                obs, reward, terminated, truncated, _ = env.step(
                    {k: v.cpu().numpy()[0] for k, v in action.items()}
                )
                done = terminated or truncated
                ep_reward += reward
                ep_length += 1
            
            rewards.append(ep_reward)
            lengths.append(ep_length)
            print(f"   Episode {ep+1}: Reward={ep_reward:.2f}, Length={ep_length}")
    
    print("="*80)
    print(f"\n📊 Results:")
    print(f"   Mean Reward: {np.mean(rewards):.2f} ± {np.std(rewards):.2f}")
    print(f"   Mean Length: {np.mean(lengths):.1f} ± {np.std(lengths):.1f}")
    print(f"   Best: {np.max(rewards):.2f} | Worst: {np.min(rewards):.2f}")
    print("="*80)
````

````python
# Evaluate on multiple episodes
eval_env = BalatroEnv(
    max_steps=config['environment']['max_steps'],
    starting_deck_size=config['environment']['starting_deck_size']
)

n_eval_episodes = config['evaluation']['n_eval_episodes']
eval_rewards = []
eval_lengths = []
eval_max_scores = []

model.eval()
print(f"Evaluating model over {n_eval_episodes} episodes...")

with torch.no_grad():
    for episode in range(n_eval_episodes):
        obs, info = eval_env.reset()
        done = False
        episode_reward = 0
        episode_length = 0
        max_score = 0
        
        while not done:
            # Convert observation to tensor
            obs_tensor = {k: torch.FloatTensor(v).unsqueeze(0).to(device) for k, v in obs.items()}
            
            # Get action (deterministic)
            action, _, _, _ = model.get_action_and_value(obs_tensor, deterministic=True)
            
            # Step environment
            obs, reward, terminated, truncated, info = eval_env.step(
                {k: v.cpu().numpy()[0] for k, v in action.items()}
            )
            
            done = terminated or truncated
            episode_reward += reward
            episode_length += 1
            max_score = max(max_score, info.get('score', 0))
        
        eval_rewards.append(episode_reward)
        eval_lengths.append(episode_length)
        eval_max_scores.append(max_score)
        
        print(f"  Episode {episode+1}/{n_eval_episodes}: "
              f"Reward={episode_reward:.2f}, Length={episode_length}, Max Score={max_score}")

print(f"\nEvaluation Results:")
print(f"  Mean Reward: {np.mean(eval_rewards):.2f} ± {np.std(eval_rewards):.2f}")
print(f"  Mean Length: {np.mean(eval_lengths):.1f} ± {np.std(eval_lengths):.1f}")
print(f"  Mean Max Score: {np.mean(eval_max_scores):.1f} ± {np.std(eval_max_scores):.1f}")
print(f"  Best Episode Reward: {np.max(eval_rewards):.2f}")
print(f"  Worst Episode Reward: {np.min(eval_rewards):.2f}")
````
