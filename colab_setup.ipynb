## ⚙️ Step 1: Check GPU

**IMPORTANT:** Runtime → Change runtime type → GPU (T4)

````python
import subprocess

print("="*70)
print("🔍 GPU DETECTION")
print("="*70)

try:
    gpu_info = subprocess.check_output(
        ['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv,noheader'],
        encoding='utf-8'
    ).strip()
    gpu_name, gpu_mem = gpu_info.split(', ')
    print(f"✓ GPU: {gpu_name}")
    print(f"✓ VRAM: {gpu_mem}")
    
    if 'T4' in gpu_name:
        print("\n🎉 Perfect! T4 detected - optimized for you!")
    elif 'A100' in gpu_name:
        print(f"\n🚀 {gpu_name}! Increase batch_size to 768 for faster training.")
    else:
        print(f"\n✓ {gpu_name} detected!")
except Exception as e:
    print("❌ NO GPU! Change runtime type to GPU")
    print(f"Error: {e}")
    import sys
    sys.exit(1)

print("="*70)
````

## 🛠️ Troubleshooting

### Out of Memory (OOM)
Edit `configs/full_jokers_colab.yaml`:
- `batch_size: 512` → `384` or `256`
- `d_model: 320` → `256`
- `gradient_checkpointing: true`

### Slow Training
Check GPU utilization:
```bash
!nvidia-smi  # Should be ~95-100%
```

### Disconnections
- Keep tab active
- Use Drive backup (Step 10)
- Download checkpoints regularly
- Consider Colab Pro
